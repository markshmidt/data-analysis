```python
import pandas as pd
import duckdb



con = duckdb.connect()
df_volunteers= pd.read_csv("../data/all_volunteers.csv")

df_hours = pd.read_csv("../data/volunteer_hour_log.csv")

con.register("volunteers", df_volunteers)
con.register("hours", df_hours)

con.execute("""
CREATE VIEW raw_volunteers AS
SELECT *
FROM read_csv_auto('../data/all_volunteers.csv', strict_mode = false);
""")

con.execute("""
CREATE VIEW raw_hours AS
SELECT *
FROM read_csv_auto('../data/volunteer_hour_log.csv', strict_mode = false);
""")

```

```python
con.execute(r"""
CREATE OR REPLACE VIEW clean_volunteers AS
WITH base AS (
  SELECT
    TRIM(Name) AS full_name,

    -- 1) extract "Sep 26, 2025 3:07 AM" from the messy text
    regexp_extract(
      "Creation Log",
      '([A-Z][a-z]{2}\s+\d{1,2},\s+\d{4}\s+\d{1,2}:\d{2}\s+[AP]M)',
      1
    ) AS date_joined_vps,

    LOWER(TRIM("Email Address")) AS email,
    

  FROM raw_volunteers
)
SELECT
  full_name,

  -- 2) parse into TIMESTAMP (safe)
  TRY_STRPTIME(date_joined_vps, '%b %d, %Y %I:%M %p') AS date_joined_vps,

  email,
FROM base
WHERE full_name IS NOT NULL
""")

df_volunteers_clean = con.execute("""
SELECT *
FROM clean_volunteers
LIMIT 20
""").df()

df_volunteers_clean
```

```python
con.execute(r"""
CREATE OR REPLACE VIEW clean_hours AS
WITH base AS (
    SELECT
        TRIM("First Name") AS first_name,
        TRIM("Your Last Name") AS last_name,
        TRIM("First Name") || ' ' || TRIM("Your Last Name") AS full_name,

        -- extract "Sep 26, 2025 3:07 AM"
        regexp_extract(
            "Creation Log",
            '([A-Z][a-z]{2}\s+\d{1,2},\s+\d{4}\s+\d{1,2}:\d{2}\s+[AP]M)',
            1
        ) AS date_joined_vps_str,

        LOWER(TRIM("Email")) AS email,

        -- already a DATE in most CSVs
        TRY_CAST("Volunteering Date" AS DATE) AS volunteering_date,

        "Total Volunteering Hours Completed Today" AS total_hours,
        "log type" AS log_type
    FROM raw_hours
)
SELECT
    first_name,
    last_name,
    full_name,
    TRY_STRPTIME(date_joined_vps_str, '%b %d, %Y %I:%M %p') AS date_joined_vps,
    email,
    volunteering_date,
    total_hours,
    log_type
FROM base
WHERE full_name IS NOT NULL;
""")


df_hours_clean = con.execute("""
SELECT *
FROM clean_hours
LIMIT 20
""").df()
df_hours_clean
```

## Question 1. In Jan 2026, who participated in volunteering and for how many hours

```python
query = """
SELECT
  v.full_name,
  v.email,
  v.date_joined_vps,

  -- January 2026 only
  SUM(CASE
        WHEN h.log_type = 'Training'
         AND h.volunteering_date >= DATE '2026-01-01'
         AND h.volunteering_date <  DATE '2026-02-01'
        THEN h.total_hours ELSE 0
      END) AS training_hours_jan_2026,

  SUM(CASE
        WHEN h.log_type = 'Service'
         AND h.volunteering_date >= DATE '2026-01-01'
         AND h.volunteering_date <  DATE '2026-02-01'
        THEN h.total_hours ELSE 0
      END) AS service_hours_jan_2026,

  -- all time totals
  SUM(CASE WHEN h.log_type = 'Training' THEN h.total_hours ELSE 0 END)
    AS training_hours_total,

  SUM(CASE WHEN h.log_type = 'Service' THEN h.total_hours ELSE 0 END)
    AS service_hours_total

FROM clean_volunteers v
LEFT JOIN clean_hours h
  ON v.email = h.email

GROUP BY
  v.email,
  v.full_name,
  v.date_joined_vps

ORDER BY
  service_hours_jan_2026 DESC,
  training_hours_jan_2026 DESC;

"""
df_jan_2026_hours = con.execute(query).df()
df_jan_2026_hours.to_excel("../data/jan_2026_volunteer_hours.xlsx", index=False)
```

## Question 2. By Dec 31, 2025, how many VSP volunteers have completed hours in each of the following range:
0-40hrs
40-80hrs
80-120hrs
Above 120hrs


```python
query = """
WITH hours_by_volunteer AS (
  SELECT
    email,
    SUM(total_hours) AS total_hours
  FROM clean_hours
  WHERE volunteering_date <= DATE '2025-12-31'
  GROUP BY email
)
SELECT
  CASE
    WHEN total_hours >= 0   AND total_hours < 40  THEN '0–40 hrs'
    WHEN total_hours >= 40  AND total_hours < 80  THEN '40–80 hrs'
    WHEN total_hours >= 80  AND total_hours < 120 THEN '80–120 hrs'
    ELSE 'Above 120 hrs'
  END AS hour_range,

  COUNT(*) AS volunteer_count

FROM hours_by_volunteer
GROUP BY hour_range
ORDER BY
  CASE hour_range
    WHEN '0–40 hrs' THEN 1
    WHEN '40–80 hrs' THEN 2
    WHEN '80–120 hrs' THEN 3
    ELSE 4
  END;
"""
df_hours_range = con.execute(query).df()
df_hours_range.to_excel("../data/volunteer_hours_range_by_dec_31_2025.xlsx", index=False)
```
